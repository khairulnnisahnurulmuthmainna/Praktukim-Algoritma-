```python
# Fungsi dengan default parameter + return
def konversi_nilai(huruf, skala="4.00"):
    nilai = {"A": 4, "B": 3, "C": 2, "D": 1, "E": 0}
    return nilai.get(huruf.upper(), None)

def hitung_rata(total=0, jumlah=0):
    if jumlah == 0:
        return 0
    return total / jumlah

total = 0
jumlah = 0

while True:
    huruf = input("masukkan nilai: ")
    if huruf == "":
        break
    angka = konversi_nilai(huruf)
    if angka is None:
        print("Huruf tidak valid, masukkan A/B/C/D/E")
        continue
    print("nilai =", angka)
    total += angka
    jumlah += 1

print("Rata-ratanya adalah: ", hitung_rata(total, jumlah))
```

```python
# Fungsi 1: cek tahun kabisat
def is_kabisat(tahun):
    return (tahun % 4 == 0 and tahun % 100 != 0) or (tahun % 400 == 0)

# Fungsi 2: hitung jumlah hari (tahun = default parameter)
def jumlah_hari(bulan, tahun=2023):
    if bulan == 2:
        return 29 if is_kabisat(tahun) else 28
    elif bulan in [4, 6, 9, 11]:
        return 30
    else:
        return 31

print("Program ini akan menentukan jumlah hari dalam bulan tertentu")

while True:
    print("Masukkan 0 untuk menghentikan program")
    bulan = int(input("masukkan bulan (1-12): "))

    if bulan == 0:
        print("Terima kasih sudah menggunakan program saya, sampai berjumpa lagi.")
        break
    elif bulan < 1 or bulan > 12:
        print("Bulan tidak valid!")
    elif bulan == 2:
        tahun = int(input("masukkan tahun (e.g., 2021): "))
        print(jumlah_hari(bulan, tahun), "hari dalam sebulan")
    else:
        print(jumlah_hari(bulan), "hari dalam sebulan")
```

```python
# Fungsi 1: cek apakah n prima
def is_prima(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

# Fungsi 2: tampilkan hasil
def tampilkan_hasil(n):
    if is_prima(n):
        print(n, "adalah bilangan prima")
    else:
        print(n, "bukan bilangan prima")

angka = int(input("Masukkan bilangan: "))
tampilkan_hasil(angka)
```

```python
# Fungsi dengan parameter, mengembalikan (angka, akhiran)
def ordinal(n):
    if 10 <= n % 100 <= 20:
        akhiran = "th"
    elif n % 10 == 1:
        akhiran = "st"
    elif n % 10 == 2:
        akhiran = "nd"
    elif n % 10 == 3:
        akhiran = "rd"
    else:
        akhiran = "th"
    return n, akhiran

print("Ordinal Number")
print("ketik 0 untuk mengentikan program")

while True:
    angka = int(input("masukkan angka: "))
    print(ordinal(angka))
    if angka == 0:
        break

print("terima kasih telah menggunakan program saya")
```
